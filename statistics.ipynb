```python

# Kurtosis
# Kurtosis tells us how much data is present in the tails of a distribution and how likely extreme values (outliers) are.

# In simple words:
# Kurtosis helps us understand whether our data has more extreme values or fewer extreme values.
# It is mainly used in statistics and data analysis to understand the shape of a data distribution

# 3types:
# 1. Mesokurtic (Normal Distribution)
# 2. Leptokurtic (High Kurtosis)
# 3. Platykurtic (Low Kurtosis)
```

```python
# 1. Mesokurtic (Normal Distribution)
import pandas as pd
data = [45, 48, 49, 50, 50, 51, 52, 55, 56]
df = pd.DataFrame({"Values": data})
df["Values"].kurt()
```

```python
# 2. Leptokurtic (High Kurtosis)
data1 = [10, 11, 12, 12, 13, 13, 14, 15, 100]
df1 = pd.DataFrame({"Values": data1})
df1["Values"].kurt()
```

```python
# 3. Platykurtic (Low Kurtosis)
data2 = [10, 20, 30, 40, 50, 60, 70, 80, 90]
df2=pd.DataFrame({"Values":data2})
df2["Values"].kurt()
```

```python
# BAYES ' THEOREM


# Think of 100 people.

# 30 are smokers.
# Among those 30 smokers, 20% get a heart attack.

# So: 30 x 20% = 6

# 6 people are smokers with heart attacks.
# Now suppose we know that 8 people in total had heart attacks

# Question: If I randomly pick one of those 8 heart-attack patients, what is the chance they are a smoker?

# Out of the 8 heart-attack patients:
# 6 are smokers
# 2 are non-smokers

# Therefore:
# P(Smoker | Heart Attack) = 6 / 8 = 0.75 = 75%

# Simple Interpretation
# Before knowing anything:
# Chance a person is a smoker = 30%

# After learning the person had a heart attack:
# Chance the person is a smoker = 75%

# Bayes' Theorem is simply:
# Start with an initial probability and update it when you get new information.

# In this case:

# Initial information:
# 30% are smokers

# New information:
# The person had a heart attack

# Updated probability:
# 75% chance they are a smoker
```

```python
#BINOMIAL DISTRIBUTION

# n=10 + 10 coin tosses
# p=0.5 + probability of heads
# size=1000 + repeat experiment 1000 times
import numpy as np 
import matplotlib.pyplot as plt
data=np.random.binomial(n=10,p=0.5,size=1000)

plt.hist(data)
plt.title("Binomial Distribution")
plt.show()
```

```python
 #POISSON DISTRIBUTION

# Poisson Distribution
# Example: Number of customers arriving per hour.
# Explanation
# lam=5 -> average 5 customers per hour
import numpy as np
import matplotlib.pyplot as plt
data=np.random.poisson(lam=5,size=1000)
plt.hist(data)
plt.title("Poisson Distribution")
plt.show
```

```python
# Exponential Distribution

# Example: Time between customer arrivals.
# Explanation
# scale=2 + average waiting time
import numpy as np
import matplotlib.pyplot as plt
data=np.random.exponential(scale=2,size=1000)
plt.hist(data)
plt.title("Poisson Distribution")
plt.show
```

```python
# #Normal Distribution

# Example: Student marks.
# Explanation
# loc=70 + mean
# scale=10 + standard deviation
# Produces the famous bell curve.
import numpy as np
import matplotlib.pyplot as plt
data=np.random.normal(loc=70,scale=10,size=1000)
plt.hist(data)
plt.title("Poisson Distribution")
plt.show
```
