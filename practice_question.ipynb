```python
# . Identify qualitative and quantitative variables from a dataset.
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "John", "Riya", "David"],
    "Age": [22, 25, 30, 28, 35],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "Department": ["IT", "HR", "Sales", "IT", "HR"],
    "Salary": [25000, 35000, 45000, 30000, 50000]
}
df = pd.DataFrame(data)
qualitative=df.select_dtypes(include="object").columns
quantitative=df.select_dtypes(include="number").columns
print("qualitative",list(qualitative))
print("quantitsative",list(quantitative))
```

```python
# . Calculate Variance and Standard Deviation.
import numpy as np
data=[10,20,30]
variance=np.var(data)
standard=np.std(data)
print("variance:",variance)
print("stanndard deviation:",standard)
```

```python
# . Create a Histogram and identify the distribution shape.
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80,
         80, 85, 90, 90, 95, 60, 65, 70, 75, 80]
df=pd.DataFrame(marks,columns=["marks"])
print("skewenes:",df["marks"].skew())

plt.hist(marks,bins=5,edgecolor="black")
plt.title("mark distribution")
plt.xlabel("marks")
plt.ylabel("number of students")
plt.show()
```

```python
#probability of even dice
dice=[1,2,3,4,5,6]
even_count=0
for i in dice:
    if i %2==0:
        even_count+=1
probability=even_count/len(dice)
print(probability)

```
